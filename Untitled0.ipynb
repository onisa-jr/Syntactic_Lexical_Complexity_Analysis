```python
import os
import pandas as pd
import subprocess

# Define paths
BASE_DIR = "/content/drive/MyDrive/syntactic_and_lexial_complexity_analysis"
DATA_DIR = os.path.join(BASE_DIR, "data")
NEOSCA_DIR = os.path.join(BASE_DIR, "neosca")

SCA_CSV = os.path.join(NEOSCA_DIR, "neosca_sca_results.csv")
LCA_CSV = os.path.join(NEOSCA_DIR, "neosca_lca_results.csv")

# Make sure directories exist
assert os.path.isdir(DATA_DIR), f"Missing data dir: {DATA_DIR}"
os.makedirs(NEOSCA_DIR, exist_ok=True)

# Run NeoSCA once for all files using wildcard pattern
print("🔁 Running NeoSCA on all .txt files...")
try:
    subprocess.run(["python", "-m", "neosca", "sca", f"{DATA_DIR}/*.txt"], check=True)
    subprocess.run(["python", "-m", "neosca", "lca", f"{DATA_DIR}/*.txt"], check=True)
    print("✅ NeoSCA analysis completed.")
except subprocess.CalledProcessError as e:
    print(f"❌ NeoSCA failed:\n{e.stderr}")
    raise

df_sca = pd.read_csv(SCA_CSV)
df_lca = pd.read_csv(LCA_CSV)

# Extract participant ID from file path
df_sca["ID"] = df_sca["Filepath"].apply(lambda x: int(os.path.splitext(os.path.basename(x))[0]))
df_lca["ID"] = df_lca["Filepath"].apply(lambda x: int(os.path.splitext(os.path.basename(x))[0]))

# Clean and order columns
df_sca = df_sca.drop(columns=["Filepath"])
df_lca = df_lca.drop(columns=["Filepath"])

df_sca = df_sca.sort_values("ID")[["ID"] + [c for c in df_sca.columns if c != "ID"]]
df_lca = df_lca.sort_values("ID")[["ID"] + [c for c in df_lca.columns if c != "ID"]]

# Export to Excel
sca_excel = os.path.join(BASE_DIR, "syntactic_complexity.xlsx")
lca_excel = os.path.join(BASE_DIR, "lexical_complexity.xlsx")

df_sca.to_excel(sca_excel, index=False)
df_lca.to_excel(lca_excel, index=False)

print(f"📦 Results saved to:\n→ {sca_excel}\n→ {lca_excel}")
```

```python
def extract_metrics(csv_path, output_dir, prefix, selected_metrics):
    """
    Extract specific metrics and save full + filtered versions.

    Args:
        csv_path (str): Path to NeoSCA-generated file (should be .xlsx).
        output_dir (str): Where to save results.
        prefix (str): File prefix ("syntactic" or "lexical").
        selected_metrics (List[str]): List of metric column names to extract.
    """
    # Read the Excel file
    df = pd.read_excel(csv_path)

    # Extract ID from Filepath (assuming Filepath column exists and contains the original file paths)
    # If Filepath column is not available in the Excel, we might need to adjust how ID is obtained.
    if "Filepath" in df.columns:
        df["ID"] = df["Filepath"].apply(lambda x: os.path.splitext(os.path.basename(x))[0])
         # Reorder, keeping original columns and adding ID
        df = df[["ID"] + [col for col in df.columns if col not in ["ID", "Filepath"]]]
    else:
        # If Filepath is not in the Excel, assume ID is already a column or handle accordingly.
        # For now, we'll assume ID is a column already if Filepath is missing.
        if "ID" not in df.columns:
             print(f"Warning: 'Filepath' and 'ID' columns not found in {os.path.basename(csv_path)}. Cannot extract/reorder by ID.")


    # Filter
    filtered_df = df[["ID"] + [col for col in selected_metrics if col in df.columns]]

    # Save both
    df.to_excel(os.path.join(output_dir, f"{prefix}_complexity_full.xlsx"), index=False)
    filtered_df.to_excel(os.path.join(output_dir, f"{prefix}_complexity_14_metrics.xlsx"), index=False)

    print(f"✅ {prefix.title()} full metrics → {prefix}_complexity_full.xlsx")
    print(f"✅ {prefix.title()} 14 selected metrics → {prefix}_complexity_14_metrics.xlsx")


# Define the 14 SCA and LCA metrics to keep
SCA_METRICS = ["MLC", "MLS", "MLT", "C/T", "CT/T", "DC/C", "DC/T",
               "CP/C", "CP/T", "T/S", "CN/C", "CN/T", "VP/T", "C/S"]

LCA_METRICS = ["LS1", "LS2", "VS1", "VS2", "CVS1", "NDW-50", "NDW-ER50",
               "NDW-ES50", "VV1", "SVV1", "CVV1", "VV2", "NV", "ModV"]


# Define paths - ensure these match the paths where the .xlsx files were saved
BASE_DIR = "/content/drive/MyDrive/syntactic_and_lexial_complexity_analysis"
sca_excel = os.path.join(BASE_DIR, "syntactic_complexity.xlsx")
lca_excel = os.path.join(BASE_DIR, "lexical_complexity.xlsx")


extract_metrics(
    csv_path=sca_excel,
    output_dir=BASE_DIR,
    prefix="syntactic",
    selected_metrics=SCA_METRICS
)

extract_metrics(
    csv_path=lca_excel,
    output_dir=BASE_DIR,
    prefix="lexical",
    selected_metrics=LCA_METRICS
)
```
